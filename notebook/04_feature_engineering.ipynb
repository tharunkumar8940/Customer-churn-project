```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
df = pd.read_csv("customer_churn_clean.csv")
```

```python
df.head()
```

```python
df.dtypes
```

```python
encoder = LabelEncoder()
```

```python
df["Gender"] = encoder.fit_transform(df["Gender"])

df["ContractType"] = encoder.fit_transform(df["ContractType"])

df["InternetService"] = encoder.fit_transform(df["InternetService"])

df["TechSupport"] = encoder.fit_transform(df["TechSupport"])

df["PaymentMethod"] = encoder.fit_transform(df["PaymentMethod"])

df["Churn"] = encoder.fit_transform(df["Churn"])
```

```python
df.head()
```

```python
X = df.drop("Churn", axis=1)

y = df["Churn"]
```

```python
print(X.shape)

print(y.shape)
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
X_scaled = pd.DataFrame(X_scaled, columns=X.columns)

X_scaled.head()
```

```python
X_scaled.describe()
```

```python
X_scaled.to_csv("X_scaled.csv", index=False)

y.to_csv("y.csv", index=False)
```
